```python
import sqlite3
import pandas as pd
import os
```

```python
# Define Paths
DATABASE_PATH = os.path.join("..", "..", "data", "f1_fantasy.db")
PROCESSED_DIR = os.path.join("..", "..", "data", "processed")

processed_files = os.listdir(PROCESSED_DIR)
```

```python
# Connect to SQLite
con = sqlite3.connect(DATABASE_PATH)
```

```python
    # Find Driver CSVs
driver_files = []

for file in processed_files:
        if file.startswith("drivers_race_"):
            driver_files.append(file)
```

```python
driver_files.sort()
driver_files
```

```python
    # Read Driver + Create Dataframe
driver_dfs = []

for file in driver_files:
        filepath = os.path.join(PROCESSED_DIR, file) 
        df = pd.read_csv(filepath)
        driver_dfs.append(df)
 
# Creates a list of dataframes
driver_dfs
```

```python
# Stack Driver CSVs
drivers_df = pd.concat(driver_dfs, ignore_index=True)
drivers_df
```

```python
drivers_df.shape
```

```python
drivers_df["race_number"].unique()
```

```python
# Write to Drivers Race Snapshots in DB
drivers_df.to_sql("driver_race_snapshots", con, if_exists="replace", index=False)
```

```python
from pathlib import Path
import json
import pandas as pd

records = []

for file_path in Path("../../data/raw/detailed_teams").glob("*.json"):
    with open(file_path) as f:
        raw = json.load(f)

    team = raw["Data"]["Value"]["userTeam"][0]

    records.append({
        "race_number": raw["Data"]["Value"]["mdid"],
        "team_no": team["teamno"],
        "team_name": team["teamname"],
        "teamval": team.get("teamval"),
        "teambal": team.get("teambal"),
        "maxteambal": team.get("maxteambal"),
    })

values_df = pd.DataFrame(records)

values_df["teamval_plus_teambal"] = (
    values_df["teamval"] + values_df["teambal"]
)

values_df.sort_values(
    ["race_number", "team_name"]
).head(30)
```

```python
check_df = values_df.dropna(
    subset=["teamval", "teambal", "maxteambal"]
).copy()

check_df["difference"] = (
    check_df["maxteambal"]
    - check_df["teamval_plus_teambal"]
).round(2)

check_df["difference"].value_counts()
```

```python
values_df[
    ["teamval", "teambal", "maxteambal"]
].isna().sum()
```

```python
values_df.groupby("race_number").agg(
    teams=("team_no", "size"),
    teamval_available=("teamval", "count"),
    maxteambal_available=("maxteambal", "count"),
)
```

## Conclusion / transformation plan

Existing processed CSVs:
- `asset_snapshot` = asset × race
- `league_standings_snapshot` = team × race
- `team_asset_snapshot` = team × race × asset

Use the detailed-team JSON to:

- **Enrich `league_standings_snapshot`** with race-specific budget/transfer data (`teambal`, `usersubs`, `usersubsleft`) and chip/captain information where useful.
- **Enrich `team_asset_snapshot`** with `iscaptain`, `ismgcaptain`, `playerpostion` and `isfinal`.
- **Create `team_chip_usage`** for the 6 possible chips and the race each was used.
- Keep season/current fields such as `ovpoints` and `ovrank` separate from race history (only create a season table if useful).

`teamval` is not reliable for historical team value because it may not update when a team is unchanged. Instead calculate:

**team value = sum of the 7 assets' values for that race + `teambal`**

using `team_asset_snapshot` + `asset_snapshot`.

Also remove `user_name` from processed data and use `team_name` instead.

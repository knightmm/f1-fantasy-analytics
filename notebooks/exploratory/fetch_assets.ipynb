```python
import requests
import pandas as pd
```

```python
# format is race number_language
# race 5 = canada
url = "https://fantasy.formula1.com/feeds/drivers/5_en.json"

r = requests.get(url)

print(r.status_code)
```

```python
data = r.json()

print(type(data))
print(data.keys())
```

```python
# dumping the json before dataframe changes
import os
import json
```

```python
race_number = 5
```

```python
save_path = os.path.join("..", "..", "data", "raw")

with open(f"{save_path}/assets_race_{race_number}.json", "w") as f:
    json.dump(data, f, indent=2)
```

```python
# appending the API feed timestamp to the dataframe containing asset data
df["feed_time_utc"] = data["Data"]["FeedTime"]["UTCTime"]
```

```python
df.head(2)
```

```python
# extracting the race number from the URL used to fetch data
import re

race_number = re.search(r"drivers/(\d+)_", url).group(1)
race_number
```

```python
#appending the race number to the dataframe
df["race_number"] = race_number
df.head(2)
```

```python
cols_to_front = ["race_number", "feed_time_utc"]
remaining_cols = [col for col in df.columns if col not in cols_to_front]

df = df[cols_to_front + remaining_cols]
df.head(1)
```

```python
df["race_number"] = df["race_number"].astype(int)
df["feed_time_utc"] = pd.to_datetime(df["feed_time_utc"], format="%m/%d/%Y %I:%M:%S %p", utc=True)
df.head(1)
```

```python
numeric_cols = ['Value', 'OverallPpints', 'GamedayPoints', 'SelectedPercentage', 
       'CaptainSelectedPercentage', 'OldPlayerValue', 'BestRaceFinished', 
       'HigestGridStart', 'HigestChampFinish', 'FastestPitstopAward',
       'BestRaceFinishCount', 'HighestGridStartCount',
       'HighestChampFinishCount', 'FastestPitstopAwardCount',
       'QualifyingPoints', 'RacePoints', 'SprintPoints', 'NoNegativePoints', 'old_Value',
       'new_value', 'ProjectedGamedayPoints', 'ProjectedNoNegativePoints',
       'ProjectedOverallPpints', 'AdditionalStats.fastest_lap_pts',
       'AdditionalStats.dotd_pts', 'AdditionalStats.overtaking_pts',
       'AdditionalStats.q3_finishes_pts',
       'AdditionalStats.top10_race_position_pts',
       'AdditionalStats.top8_sprint_position_pts',
       'AdditionalStats.total_position_pts',
       'AdditionalStats.total_position_gained_lost',
       'AdditionalStats.total_dnf_dq_pts', 'AdditionalStats.value_for_money']

```

```python
int_cols = [
    "GamedayPoints",
    "BestRaceFinished",
    "HigestGridStart",
    "BestRaceFinishCount",
    "HighestGridStartCount",
    "HighestChampFinishCount",
    "FastestPitstopAwardCount",
]

float_cols = [
    "Value",
    "OverallPpints",
    "QualifyingPoints",
    "RacePoints",
    "SprintPoints",
    "NoNegativePoints",
    "ProjectedGamedayPoints",
    "ProjectedNoNegativePoints",
    "ProjectedOverallPpints",
    "SelectedPercentage",
    "CaptainSelectedPercentage",
    "OldPlayerValue",
    "FastestPitstopAward",
    "old_Value",
    "new_value",
    "HigestChampFinish",
]
```

```python
for col in int_cols:
    df[col] = pd.to_numeric(df[col], errors="coerce").astype("Int64")

for col in float_cols:
    df[col] = pd.to_numeric(df[col], errors="coerce")
```

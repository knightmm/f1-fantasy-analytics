# Drivers

```python
import requests
import pandas as pd
```

```python
# format is race number_language
# race 5 = canada
url = "https://fantasy.formula1.com/feeds/drivers/5_en.json"

r = requests.get(url)

print(r.status_code)
```

```python
data = r.json()

print(type(data))
print(data.keys())
```

```python
print(type(data["Data"]))
```

```python
print(data["Data"].keys())
```

```python
drivers = data["Data"]["Value"]
```

```python
print(type(drivers))
print(len(drivers))
print(drivers[0])
```

```python
df = pd.json_normalize(drivers)

df.head()
```
